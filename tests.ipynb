```python
import random, math

rng = random.Random(7)

n_points = 8
n_axis = math.ceil(math.sqrt(n_points))

candidate_cells = [
        (ix, iy)
        for ix in range(n_axis)
        for iy in range(n_axis)
    ]

selected_cells = rng.sample(candidate_cells, n_points)

print(selected_cells)
```
